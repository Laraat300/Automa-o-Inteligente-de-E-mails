Objetivo

Ler uma lista de clientes de um arquivo exel e enviar um e-mail para cada um.

Requisitos
Ler arquivo clientes.csv.
Personalizar o nome do destinatário.
Enviar e-mails usando smtplib.

```python
import pandas as pd, pyautogui, time

tabela = pd.read_csv("email.csv")
print (tabela)
```

```python
import pyperclip
pyautogui.PAUSE = 0.5
pyautogui.press("win")
pyautogui.write("email")
pyautogui.press("enter")
time.sleep(5)   

for linha in tabela.index:
    pyautogui.click(x=140, y=105)
    email = tabela.loc[linha, "Email"]
    pyautogui.write(str(email))
    pyautogui.press("enter")
    pyautogui.press("tab")
    pyautogui.write("Promocao!!")
    pyautogui.press("tab")
    nome = tabela.loc[linha, "Nome"]
    mensagem = f"""OLÁ {nome}
    🎉 SUPER PROMOÇÃO IMPERDÍVEL! 🎉

    Aproveite nossa Semana de Ofertas e garanta descontos incríveis em toda a loja!

    ✅ Até 50% de desconto em produtos selecionados
    ✅ Frete grátis para compras acima de R$ 199,00
    ✅ Parcelamento em até 10x sem juros
    ✅ Brinde exclusivo nas primeiras 100 compras

    ⏰ Promoção válida por tempo limitado!

    Não perca a oportunidade de economizar e levar para casa os produtos que você deseja com condições especiais.

    📲 Acesse nosso site ou visite nossa loja e aproveite agora mesmo!

    Sua satisfação é a nossa prioridade. Venha conferir! 💙

    Consulte regulamento e condições da promoção.
    """
    pyperclip.copy(mensagem)
    pyautogui.hotkey("ctrl", "v")
    pyautogui.click(x=701, y=157)
```
